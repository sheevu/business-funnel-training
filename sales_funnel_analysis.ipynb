## 4. Stage Conversion Rate Analysis
Modeling conversion efficiency across key funnel transitions.

```python
stages = ["Awareness -> Lead", "Lead -> Prospect", "Prospect -> Close", "Close -> Repeat"]
conversion_rates = [12, 28, 15, 40]

plt.figure(figsize=(9, 5))
bars = plt.barh(stages, conversion_rates, color="#2563eb", edgecolor="#1e40af", height=0.55)
plt.xlabel("Conversion Percentage (%)", fontsize=12, fontweight="bold")
plt.title("Funnel Stage Conversion Rate Benchmarks", fontsize=14, fontweight="bold")
for bar in bars:
    width = bar.get_width()
    plt.text(width + 0.5, bar.get_y() + bar.get_height()/2, f"{width}%", va="center", fontweight="bold")
plt.xlim(0, 50)
plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()
```

## 5. Revenue Velocity & Compound CLV Growth Projection
Simulating quarterly revenue velocity driven by Up-Sale & Repeat business.

```python
quarters = ["Q1", "Q2", "Q3", "Q4"]
retention_revenue = [100, 140, 210, 320]

plt.figure(figsize=(8, 4.5))
plt.plot(quarters, retention_revenue, marker="o", color="#fb923c", linewidth=3, markersize=8, label="Retention Revenue Index")
plt.fill_between(quarters, retention_revenue, color="#fb923c", alpha=0.2)
plt.title("Revenue Velocity Projection (Retention & Up-Sell)", fontsize=14, fontweight="bold")
plt.ylabel("Revenue Index (Base=100)", fontsize=12)
plt.grid(True, linestyle="--", alpha=0.5)
for i, txt in enumerate(retention_revenue):
    plt.annotate(f"{txt} pts", (quarters[i], retention_revenue[i] + 8), ha="center", fontweight="bold")
plt.ylim(80, 360)
plt.tight_layout()
plt.show()
```

## 6. Interactive Business Funnel Simulator
Run customized scenario simulations based on traffic, lead conversion, deal size, and repeat up-sale rates.

```python
def run_funnel_simulation(monthly_traffic=10000, avg_deal_size=1500):
    awareness = monthly_traffic
    leads = awareness * 0.12
    prospects = leads * 0.28
    closed_sales = prospects * 0.15
    repeat_clients = closed_sales * 0.40
    
    initial_revenue = closed_sales * avg_deal_size
    upsell_revenue = repeat_clients * (avg_deal_size * 0.60)
    total_revenue = initial_revenue + upsell_revenue
    
    print("="*50)
    print(f"   FUNNEL SIMULATION RESULTS (Traffic: {monthly_traffic:,})")
    print("="*50)
    print(f"1. Stage 1 Awareness:     {awareness:,} Visitors")
    print(f"2. Stage 2 Leads:         {int(leads):,} Generated")
    print(f"3. Stage 3 Prospects:     {int(prospects):,} Qualified")
    print(f"4. Stage 5 Closed Sales:  {int(closed_sales):,} Deals")
    print(f"5. Stage 8 Up-Sale:       {int(repeat_clients):,} Repeat Clients")
    print("-"*50)
    print(f"💰 Initial Revenue:      ${initial_revenue:,.2f}")
    print(f"📈 Up-Sale Revenue:       ${upsell_revenue:,.2f}")
    print(f"🚀 Total Projected Rev:   ${total_revenue:,.2f}")
    print("="*50)

# Execute default simulation
run_funnel_simulation(10000, 1500)
```
